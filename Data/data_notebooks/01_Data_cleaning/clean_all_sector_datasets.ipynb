# **Clean all the Sector Datasets**

```python
import pandas as pd
import os
```

### 1. **List of all sector data files**

```python
files = [r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY50_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_REALTY_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_PHARMA_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_METAL_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_IT_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_FMCG_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_ENERGY_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_BANK_data.csv",
         r"E:\BDA\Summer 2026\Data\raw_data\sector_raw_data\NIFTY_AUTO_data.csv",
         ]
```

### 2. **Clean every files**

```python
output_folder = r"E:\BDA\Summer 2026\Data\cleaned_data\sector_cleaned_data"
for file in files:
    # Read CSV file 
    df = pd.read_csv(file,header=[0,1])
    # Flatten MultiIndex
    df.columns = df.columns.get_level_values(0)
    # Reset index
    df = df.reset_index()
    # Rename first column to Date
    df.rename(columns={"Price": "Date"}, inplace=True)

    # Convert Date column
    df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d",errors="coerce")


    # Sort by date
    df = df.sort_values("Date")

    #Remove duplicates 
    df = df.drop_duplicates()

    #Remove rows where "Date is missing"
    df = df.dropna(axis=0, how='any')# drop rows with missing data
    df = df.loc[df.index.notna()]       # keep only rows with valid Date index

    # Reset index
    df = df.reset_index(drop=True)



    #display basic information about the dataframe
    print(f"File: {file}")
    print(df.info())
    print(df.isnull().sum())

    # save cleaned file
    name = os.path.splitext(file)[0]

    output_file = os.path.join(
    output_folder,
    f"{name}_clean.csv")
    df.to_csv(output_file, index=False)

print("All sector datasets cleaned successfully!")
```

# **Mecro Data Cleaning**

```python
import pandas as pd
import os
```

```python
input_folder = r"E:\BDA\Summer 2026\Data\raw_data\macro_raw"
output_folder = r"E:\BDA\Summer 2026\Data\cleaned_data\macro_cleaned_data"

os.makedirs(output_folder, exist_ok=True)

for file in os.listdir(input_folder):
    file_path = os.path.join(input_folder, file)
    # Read Yahoo Finance CSV
    df = pd.read_csv(file_path, header=[0,1])

    # Flatten MultiIndex columns
    df.columns = df.columns.get_level_values(0)

    # Make Date a normal column
    df = df.reset_index()

    # Rename index column to Date
    df.rename(columns={"index": "Date"}, inplace=True)

    # Remove unwanted index column if present
    df.drop(columns=["index"], errors="ignore", inplace=True)


    #convert Date
    df["Date"] = pd.to_datetime(df["Date"],format="%Y-%m-%d", errors="coerce" )
    # Remove duplicate rows
    df.drop_duplicates()

    # Sort by date
    df.sort_values("Date")

    # Remove missing rows
    df.dropna()

    # Remove unwanted index column
    df = df.drop(columns=["index"], errors="ignore")

    # Reset index
    df = df.reset_index(drop=True)

    print(file)
    print(df.info())
    print(df.isnull().sum())

    name = os.path.splitext(file)[0]

    output_file = os.path.join(
    output_folder,
    f"{name}_clean.csv")

    df.to_csv(output_file, index=False)

print("Macro datasets cleaned!")
```

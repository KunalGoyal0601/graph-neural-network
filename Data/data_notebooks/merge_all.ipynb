```python
import pandas as pd
from datetime import datetime
pd.set_option("display.max_columns", None)
```

```python
sector = pd.read_csv(r"E:\BDA\Summer 2026\Data\featured_data\featured_sector_data.csv")
macro = pd.read_csv(r"E:\BDA\Summer 2026\Data\featured_data\featured_macro_data.csv")
gdelt = pd.read_csv(r"E:\BDA\Summer 2026\Data\featured_data\featured_gdelt_data.csv")

sector["Date"] = pd.to_datetime(sector["Date"])
macro["Date"] = pd.to_datetime(macro["Date"])
gdelt["Date"] = pd.to_datetime(gdelt["Date"])

master = sector.merge(
    macro,
    on="Date",
    how="inner"
)

master = master.merge(
    gdelt,
    on="Date",
    how="inner"
)

master.to_csv(
    "master_dataset.csv",
    index=False
)
```

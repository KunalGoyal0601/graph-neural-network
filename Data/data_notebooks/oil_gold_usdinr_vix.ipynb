# **Oil Gold USDINR Data**

```python
import yfinance as yf
from datetime import datetime 
import pandas as pd
```

```python
START = "2018-01-01"
END = datetime.today().strftime("%Y-%m-%d")

# India VIX
vix = yf.download("^INDIAVIX",
                  start=START,
                  end=END)
#Print the shape of the dataframe for each sector
print(f"VIX: {vix.shape}")

#Print the first few rows of the dataframe for each sector
print(f"VIX:\n{vix.head()}\n")

#Save individual csv

csv_path = r"E:\BDA\Summer 2026\Data\raw_data\vix_data.csv" 
vix.to_csv(csv_path)

print(f"Saved VIX data to {csv_path}\n")
print(f"Shape:{vix.shape}")
```

```python
# Gold
gold = yf.download("GC=F",
                   start=START,
                   end=END)

#Print the shape of the dataframe for each sector
print(f"Gold: {gold.shape}")

#Print the first few rows of the dataframe for each sector
print(f"Gold:\n{gold.head()}\n")

#Save individual csv

csv_path = r"E:\BDA\Summer 2026\Data\raw_data\gold_data.csv" 
gold.to_csv(csv_path)

print(f"Saved Gold data to {csv_path}\n")
print(f"Shape:{gold.shape}")
```

```python
# Brent Oil
oil = yf.download("BZ=F",
                  start=START,
                  end=END)

#Print the shape of the dataframe for each sector
print(f"Oil: {oil.shape}")

#Print the first few rows of the dataframe for each sector
print(f"Oil:\n{oil.head()}\n")

#Save individual csv

csv_path = r"E:\BDA\Summer 2026\Data\raw_data\oil_data.csv" 
oil.to_csv(csv_path)

print(f"Saved Oil data to {csv_path}\n")
print(f"Shape:{oil.shape}")
```

```python
# USDINR
usd_inr = yf.download("INR=X",
                      start=START,
                      end=END)
#Print the shape of the dataframe for each sector
print(f"USDINR: {usd_inr.shape}")

#Print the first few rows of the dataframe for each sector
print(f"USDINR:\n{usd_inr.head()}\n")

#Save individual csv

csv_path = r"E:\BDA\Summer 2026\Data\raw_data\usd_inr_data.csv" 
usd_inr.to_csv(csv_path)

print(f"Saved USDINR data to {csv_path}\n")
print(f"Shape:{usd_inr.shape}")
```

## **Return Calculation**

```python
vix["VIX_Return"] = vix["Close"].pct_change()

gold["Gold_Return"] = gold["Close"].pct_change()

oil["Oil_Return"] = oil["Close"].pct_change()

usd_inr["FX_Return"] = usd_inr["Close"].pct_change()
```

# **Gdelt Data Combining**

```python
import pandas as pd
import numpy as np
from datetime import datetime
```

### **Load all GDELT files**

```python
conflict = pd.read_csv(r"E:\BDA\Summer 2026\Data\cleaned_data\geo_cleaned_data\gdelt_country_conflict_clean.csv")
events = pd.read_csv(r"E:\BDA\Summer 2026\Data\cleaned_data\geo_cleaned_data\gdelt_country_events_clean.csv")
tone = pd.read_csv(r"E:\BDA\Summer 2026\Data\cleaned_data\geo_cleaned_data\gdelt_country_tone_clean.csv")
daily = pd.read_csv(r"E:\BDA\Summer 2026\Data\cleaned_data\geo_cleaned_data\gdelt_daily_2018_2026_clean.csv")
master = pd.read_csv(r"E:\BDA\Summer 2026\Data\cleaned_data\geo_cleaned_data\gdelt_master_features_clean.csv")
```

#### **Convert dates**

```python
for df in [conflict, events, tone,master]:
    df["Date"] = pd.to_datetime(df["Date"])
```

### **Merge them one by one**

why did not take daily =because daily and master has same columns

```python
gdelt = conflict.merge(events, on="Date", how="inner")

gdelt = gdelt.merge(tone, on="Date", how="inner")

gdelt = gdelt.merge(master, on="Date", how="inner")
```

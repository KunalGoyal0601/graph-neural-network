# **Combine all clean macro data**

```python
import pandas as pd
import numpy as np
import os
```

```python
input_folder = r"E:\BDA\Summer 2026\Data\cleaned_data\macro_cleaned_data"

combined_macro = pd.DataFrame()

for file in os.listdir(input_folder):

    if file.endswith(".csv"):

        path = os.path.join(input_folder, file)

        df = pd.read_csv(path)

        # Move dates from Price to Date
        df["Date"] = df["Price"]

        # Remove the old Price column
        df = df.drop(columns=["Price"])

        df = df[df["Date"] != "Date"]


        df["Date"] = pd.to_datetime(df["Date"])

        df = df.reset_index(drop=True)

        # Remove filename parts
        macro = os.path.splitext(file)[0]
        macro = macro.replace("_data_clean", "")

        if combined_macro.empty:
            combined_macro["Date"] = df["Date"]

        combined_macro[macro] = df["Close"]

combined_macro.head()
```

## **Save combain**

```python
combined_macro.to_csv(
    r"E:\BDA\Summer 2026\Data\featured_data\combined_macro_close_prices.csv",
    index=False
)
```

### **Feature Engineering**

#### **Load combined mecro data**

```python
df_macro = pd.read_csv(
    r"E:\BDA\Summer 2026\Data\featured_data\combined_macro_close_prices.csv"
)

df_macro["Date"] = pd.to_datetime(df_macro["Date"])

macro_columns = df_macro.columns.drop("Date")
```

```python
df_macro
```

#### **Calculate return**

```python
for macro in macro_columns:

    df_macro[f"{macro}_Return"] = (
        df_macro[macro].pct_change(fill_method=None)
    )
```

```python
df_macro
```

#### **Create 30-Day Moving Average**

```python
for macro in macro_columns:

    df_macro[f"{macro}_MA30"] = (
        df_macro[macro]
        .rolling(30)
        .mean()
    )
```

```python
df_macro
```

#### **Remove Missing Values**

```python
df_macro = df_macro.dropna().reset_index(drop=True)
```

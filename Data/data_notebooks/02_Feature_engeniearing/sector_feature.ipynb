# **Combine all clean sector data based on the close price**

```python
import pandas as pd
import numpy as np
import os
```

```python
import pandas as pd
import os

input_folder = r"E:\BDA\Summer 2026\Data\cleaned_data\sector_cleaned_data"

combined_df = pd.DataFrame()

for file in os.listdir(input_folder):

    if file.endswith(".csv"):

        path = os.path.join(input_folder, file)

        df = pd.read_csv(path)

        df["Date"] = pd.to_datetime(df["Date"])

        sector = os.path.splitext(file)[0]
        sector = sector.replace("_data_clean", "")


        if combined_df.empty:
            combined_df["Date"] = df["Date"]

        combined_df[sector] = df["Close"]

combined_df
```

```python
combined_df.to_csv(
    r"E:\BDA\Summer 2026\Data\featured_data\combined_sector_close_prices.csv",
    index=False
)
```

## 3. **Feature Engineering**

```python

df_sector = pd.read_csv(
    r"E:\BDA\Summer 2026\Data\featured_data\combined_sector_close_prices.csv"
)

df_sector["Date"] = pd.to_datetime(df_sector["Date"])

sector_columns = df_sector.columns.drop("Date")
```

```python
for col in df_sector.columns:
    print(col)
```

### **Return**

```python
for sector in sector_columns:

    df_sector[f"{sector}_Return"] = (
        df_sector[sector].pct_change(fill_method=None)
    )
```

```python
df_sector
```

## **Calculate log_return**

```python
for sector in sector_columns:

    df_sector[f"{sector}_Log_Return"] = np.log(
        df_sector[sector] /
        df_sector[sector].shift(1)
    )
```

```python
df_sector
```

## **Moving Average**

```python
for sector in sector_columns:

    df_sector[f"{sector}_MA30"] = (
        df_sector[sector]
        .rolling(30)
        .mean()
    )
```

## **Volatility**

```python
for sector in sector_columns:

    df_sector[f"{sector}_Volatility"] = (
        df_sector[f"{sector}_Return"]
        .rolling(30)
        .std()
    )
```

## **Momentum**

```python
for sector in sector_columns:

    df_sector[f"{sector}_Momentum"] = (
        df_sector[sector] /
        df_sector[sector].shift(10)
    ) - 1
```

## **Relative Strength**

```python
for sector in sector_columns:

    df_sector[f"{sector}_RS"] = (
        df_sector[sector] /
        df_sector["NIFTY50"]
    )
```

## **RSI**

```python
for sector in sector_columns:

    delta = df_sector[sector].diff()

    gain = delta.where(delta > 0, 0)

    loss = -delta.where(delta < 0, 0)

    avg_gain = gain.rolling(14).mean()

    avg_loss = loss.rolling(14).mean()

    rs = avg_gain / avg_loss

    df_sector[f"{sector}_RSI"] = 100 - (100/(1+rs))
```

## **Drawdown**

```python
for sector in sector_columns:

    rolling_max = df_sector[sector].cummax()

    df_sector[f"{sector}_Drawdown"] = (
        df_sector[sector]-rolling_max
    )/rolling_max
```

## **Price / MA30**

```python
for sector in sector_columns:

    df_sector[f"{sector}_Price_MA30"] = (
        df_sector[sector] /
        df_sector[f"{sector}_MA30"]
    )
```

## **Remove NaN**

```python
df_sector = (
    df_sector
    .dropna()
    .reset_index(drop=True)
)
```

## **Save**

```python
df_sector.to_csv(
    r"E:\BDA\Summer 2026\Data\featured_data\featured_sector_data.csv",
    index=False
)
```

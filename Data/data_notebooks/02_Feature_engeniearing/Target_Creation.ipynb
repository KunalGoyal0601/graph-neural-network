# **Create Up/Down targets for all sectors.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load master dataset
df = pd.read_csv(r"E:\BDA\Summer 2026\Data\featured_data\master_dataset.csv")
df
```

```python
df.columns.to_list()
```

```python
# Creating target variable
sector_columns = [
    "NIFTY50",
    "NIFTY_AUTO",
    "NIFTY_BANK",
    "NIFTY_ENERGY",
    "NIFTY_FMCG",
    "NIFTY_IT",
    "NIFTY_METAL",
    "NIFTY_PHARMA",
    "NIFTY_REALTY"
]

for sector in sector_columns:
    df[f"{sector}_Target"] = (df[f"{sector}_Return"].shift(-1) > 0).astype(int)
```

```python
df
```

```python
# View target columns
target_columns = [f"{sector}_Target" for sector in sector_columns]
df[target_columns].head()
```

```python
df[target_columns].shape
```

```python
# value counts
for target in target_columns:
    print(df[target].value_counts())
```

```python
#Visualize Target Distribution
fig, axes = plt.subplots(3,3, figsize=(15,12))

axes = axes.flatten()

for i, target in enumerate(target_columns):

    sns.countplot(
        data=df,
        x=target,
        ax=axes[i]
    )

    axes[i].set_title(target)

plt.tight_layout()

plt.show()
```

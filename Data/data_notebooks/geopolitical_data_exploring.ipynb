"gdelt_daily_2018_2026.csv"
SELECT
  DATE(PARSE_DATE('%Y%m%d', CAST(SQLDATE AS STRING))) AS Date,

  AVG(AvgTone) AS AvgTone,

  AVG(GoldsteinScale) AS AvgGoldstein,

  COUNT(*) AS EventCount,

  SUM(NumMentions) AS TotalMentions

FROM `gdelt-bq.gdeltv2.events`

WHERE SQLDATE BETWEEN 20180101 AND 20260622

GROUP BY Date

ORDER BY Date

"gdelt_master_features.csv"
SELECT
  DATE(PARSE_DATE('%Y%m%d', CAST(SQLDATE AS STRING))) AS Date,

  AVG(AvgTone) AS AvgTone,
  AVG(GoldsteinScale) AS AvgGoldstein,

  COUNT(*) AS EventCount,
  SUM(NumMentions) AS TotalMentions,

  SUM(CASE WHEN EventRootCode='14' THEN 1 ELSE 0 END) AS ProtestCount,
  SUM(CASE WHEN EventRootCode='13' THEN 1 ELSE 0 END) AS ThreatCount,
  SUM(CASE WHEN EventRootCode='15' THEN 1 ELSE 0 END) AS MilitaryCount,
  SUM(CASE WHEN EventRootCode='18' THEN 1 ELSE 0 END) AS AssaultCount,
  SUM(CASE WHEN EventRootCode='19' THEN 1 ELSE 0 END) AS FightCount,
  SUM(CASE WHEN EventRootCode='20' THEN 1 ELSE 0 END) AS MassViolenceCount,

  SUM(CASE WHEN EventRootCode IN ('03','04','05') THEN 1 ELSE 0 END)
      AS DiplomacyCount,

  SUM(CASE WHEN EventRootCode IN ('06','07','08') THEN 1 ELSE 0 END)
      AS CooperationCount

FROM `gdelt-bq.gdeltv2.events`

WHERE SQLDATE BETWEEN 20180101 AND 20260622

GROUP BY Date
ORDER BY Date

"gdelt_country_events.csv"
SELECT
 DATE(PARSE_DATE('%Y%m%d', CAST(SQLDATE AS STRING))) AS Date,

 SUM(CASE WHEN Actor1CountryCode='USA'
      OR Actor2CountryCode='USA'
      THEN 1 ELSE 0 END) AS USA_Events,

 SUM(CASE WHEN Actor1CountryCode='CHN'
      OR Actor2CountryCode='CHN'
      THEN 1 ELSE 0 END) AS China_Events,

 SUM(CASE WHEN Actor1CountryCode='RUS'
      OR Actor2CountryCode='RUS'
      THEN 1 ELSE 0 END) AS Russia_Events,

 SUM(CASE WHEN Actor1CountryCode='PAK'
      OR Actor2CountryCode='PAK'
      THEN 1 ELSE 0 END) AS Pakistan_Events,

 SUM(CASE WHEN Actor1CountryCode='ISR'
      OR Actor2CountryCode='ISR'
      THEN 1 ELSE 0 END) AS Israel_Events,
 
 SUM(CASE WHEN Actor1CountryCode='IND'
      OR Actor2CountryCode='IND'
      THEN 1 ELSE 0 END) AS India_Events,

 SUM(CASE WHEN Actor1CountryCode='IRN'
      OR Actor2CountryCode='IRN'
      THEN 1 ELSE 0 END) AS Iran_Events

FROM `gdelt-bq.gdeltv2.events`
WHERE SQLDATE BETWEEN 20180101 AND 20260622
GROUP BY Date
ORDER BY Date

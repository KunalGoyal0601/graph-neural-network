# **Data Preprocessing**
#### **Prepare the data for machine learning and Gnn**

```python
# Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
#Load Data
df = pd.read_csv(r"E:\BDA\Summer 2026\Data\featured_data\master_with_target.csv")
df["Date"] = pd.to_datetime(df["Date"])
df.head()
```

```python
#Data info
print(df.info())
print(df.shape)
```

```python
df.columns.to_list()
```

```python
# Define Target Columns
target_columns = [
    "NIFTY50_Target",
    "NIFTY_AUTO_Target",
    "NIFTY_BANK_Target",
    "NIFTY_ENERGY_Target",
    "NIFTY_FMCG_Target",
    "NIFTY_IT_Target",
    "NIFTY_METAL_Target",
    "NIFTY_PHARMA_Target",
    "NIFTY_REALTY_Target"
]
```

```python
# Select Features X
X = df.drop(columns=["Date"] + target_columns)
print(X.shape)
X.head()
```

```python
#Select target Y
Y = df[target_columns]
print(Y.shape)
Y.head()
```

```python
# Check for Missing values
print(X.isnull().sum().sum())
print(Y.isnull().sum().sum())
```

```python
# Train Test Split (not shuffling time-series data)
split = int(len(df)*0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

Y_train = Y.iloc[:split]
Y_test = Y.iloc[split:]
```

```python
# Verifiey traning and testing pariods
print("Training Period:")
print(df.iloc[:split]["Date"].min())
print(df.iloc[:split]["Date"].max())

print("\nTesting Period:")
print(df.iloc[split:]["Date"].min())
print(df.iloc[split:]["Date"].max())
```

```python
# Check X_train,X_test,Y_train,Y_test
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
# Feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

- my target Is already standerdize if I further scaled the values not longer become up or down 
- x_train = fit_transform (learns and then transform) , x_test=transform(took learned value then transform)


```python
X_train_scaled
```

```python
# Convert back to data frame
X_train_scaled = pd.DataFrame(X_train_scaled,
                              columns=X.columns,
                              index=X_train.index)
X_test_scaled = pd.DataFrame(X_test_scaled,
                              columns=X.columns,
                              index=X_test.index)
```

```python
X_train_scaled.head()
```

```python
output_folder = r"E:\BDA\Summer 2026\Preprossing"

X_train_scaled.to_csv(
    f"{output_folder}\\X_train.csv",
    index=False
)

X_test_scaled.to_csv(
    f"{output_folder}\\X_test.csv",
    index=False
)

Y_train.to_csv(
    f"{output_folder}\\Y_train.csv",
    index=False
)

Y_test.to_csv(
    f"{output_folder}\\Y_test.csv",
    index=False
)

print("Processed datasets saved successfully!")
```

1. Load the master data with target variable
2. split master data in X (all the features escapt Date and Target columns) & Y (target features)
3. split data in X_train,X_test,Y_train,Y_test in 80,20
4. Standard Scale X_train,X_test
5. conver back to dataframe
6. Save all the dataframes into csv files

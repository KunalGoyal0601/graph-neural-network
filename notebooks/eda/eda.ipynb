# 1. **Sector Analysis Section**

#### **Define sector columns group**

```python
sector_price_columns = [
    "NIFTY50",
    "NIFTY_AUTO",
    "NIFTY_BANK",
    "NIFTY_ENERGY",
    "NIFTY_FMCG",
    "NIFTY_IT",
    "NIFTY_METAL",
    "NIFTY_PHARMA",
    "NIFTY_REALTY"
]

sector_return_columns = [f"{col}_Return" for col in sector_price_columns]

sector_volatility_columns = [f"{col}_Volatility" for col in sector_price_columns]

sector_rsi_columns = [f"{col}_RSI" for col in sector_price_columns]

sector_drawdown_columns = [f"{col}_Drawdown" for col in sector_price_columns]

sector_momentum_columns = [f"{col}_Momentum" for col in sector_price_columns]
```

#### 1.1 **Sector Price Trends** 

```python
fig,axes = plt.subplots(3,3,figsize=(18,12))
axes = axes.flatten()
for i,sector in enumerate(sector_price_columns):
    axes[i].plot(df["Date"],df[sector])
    axes[i].set_title(sector)
    axes[i].set_xlabel("Date")
    axes[i].set_ylabel("Price")
    axes[i].grid(True)

    
plt.tight_layout()
plt.show()
```

**Observations** -> 
"During march 2020 due to covid all sectors crash then 2020-2021 recover currently in 2026 almost all sectores fall"
- "Bank rises from 25,000 to 60,000, maintaining a strong long-term uptrend (Auto and Metal also post substantial gains after 2020.)"
- "Realty shows the sharpest percentage declines during COVID"
- "FMCG experiences relatively small drawdowns and smooth, consistent growth(Pharma also remains comparatively stable with moderate volatility.)"

### **1.2 Individual Sector Trends**

```python
for sector in sector_price_columns:

    plt.figure(figsize=(12,4))

    plt.plot(df["Date"], df[sector])

    plt.title(f"{sector} Trend")

    plt.xlabel("Date")

    plt.ylabel("Index")

    plt.grid(True)

    plt.show()
```

```python
df[sector_price_columns].describe().T
```

- FMCG sector has the highest average value -> 41580

## 1.3 **Daily Return Distribution**

```python
for sector in sector_return_columns:

    plt.figure(figsize=(7,4))

    sns.histplot(df[sector], bins=30, kde=True)

    plt.title(f"{sector} Distribution")

    plt.show()
```

X-axis: Daily Return % ,Y-axis: Frequency (Number of trading days with returns in that range.)

**Observations**->""

- Every sector's return distribution is centered very close to 0.(Most trading days involve only small price movements.)
- All sectors contain a small number of extreme observations.(Market is relatively stable.
Most daily returns are between -1% and +1%.)

## **1.5 Boxplot of Returns**

```python
plt.figure(figsize=(15,6))
sns.boxplot(data = df[sector_return_columns])
plt.xticks(rotation = 45)
plt.title("Sector Return Boxplots")
plt.show()
```

y-axis represents the daily return

**Boxplot of Returns Observations** -> "All medians lie very close to zero.(On a typical trading day, daily returns are small)"
- Every sector has numerous outliers on both sides.(These correspond to days with unusually large gains or losses)

## **1.5 Volatility Comparison**

"x-axis represents the average daily volatility of each sector."

## **1.7 RSI Analysis**

```python
for sector in sector_rsi_columns:

    plt.figure(figsize=(12,4))

    plt.plot(df["Date"], df[sector])

    plt.axhline(70, color="red", linestyle="--")

    plt.axhline(30, color="green", linestyle="--")

    plt.title(f"{sector}")

    plt.show()
```

**Observations** -> Y-axis: RSI Value (0–100)
"All sectors spend most of their time in the neutral zone."
- NIFTY_METAL,NIFTY Realty frequently become overbought (>70)
- NIFTY_METAL,NIFTY Realty frequently become oversold (<30)

## **1.8 Drawdown Analysis**

```python
avg_drawdown = df[sector_drawdown_columns].min()

avg_drawdown.sort_values().plot(
    kind="barh",
    figsize=(10,6)
)

plt.title("Maximum Drawdown by Sector")

plt.show()
```

"Drawdown is how much an investor would have lost if they bought at the peak and sold at the lowest point before recovery."

**Observation** ->"x-axis shows the maximum drawdown value.The farther left the bar extends, the larger the historical loss experienced by that sector."

- NIFTY_METAL suffered the deepest losses

# 3.**GDELT Analysis Section**

## 3.1 **Select GDELT Columns**

```python
conflict_columns = [
    "USA_Conflict",
    "India_Conflict",
    "China_Conflict",
    "Russia_Conflict",
    "Pakistan_Conflict",
    "Israel_Conflict",
    "Iran_Conflict"
]

event_columns = [
    "USA_Events",
    "China_Events",
    "Russia_Events",
    "India_Event",
    "Pakistan_Events",
    "Israel_Events",
    "Iran_Events"
]

tone_columns = [
    "USA_Tone",
    "China_Tone",
    "Russia_Tone",
    "Pakistan_Tone",
    "Israel_Tone",
    "India_Tone",
    "Iran_Tone"
]

master_columns = [
    "AvgTone",
    "AvgGoldstein",
    "EventCount",
    "TotalMentions"
]

event_type_columns = [
    "ProtestCount",
    "ThreatCount",
    "MilitaryCount",
    "AssaultCount",
    "FightCount",
    "MassViolenceCount",
    "DiplomacyCount",
    "CooperationCount"
]
```

## **8.2 Summary Statistics**

```python
df[
    conflict_columns +
    master_columns +
    event_type_columns
].describe().T
```

### **3.3 Country Conflict Trends**

```python
fig, axes = plt.subplots(4,2, figsize=(18,12), sharex=True)

axes = axes.flatten()

for i, country in enumerate(conflict_columns):

    axes[i].plot(df["Date"], df[country])

    axes[i].set_title(country)

    axes[i].grid(True)
    axes[i].tick_params(axis='x', labelbottom=True)

# Remove empty subplot
fig.delaxes(axes[-1])

plt.tight_layout()

plt.show()
```

Observed:

- Frequent spikes: USA, Russia, Israel, and Iran.
- Relatively stable: India and China.
- Mostly stable with occasional extreme spikes: Pakistan.(Why -> 1.GDELT counts media-reported events,2.graph shows daily event counts, not conflict intensity)

## **8.4 Total Conflict Comparison**

## **3.5 Country Event Trends**

# 4. Cross-Domain Analysis

### **4.1 Sector vs Macro Correlation**

```python
sector_return_columns = [
    "NIFTY50_Return",
    "NIFTY_AUTO_Return",
    "NIFTY_BANK_Return",
    "NIFTY_ENERGY_Return",
    "NIFTY_FMCG_Return",
    "NIFTY_IT_Return",
    "NIFTY_METAL_Return",
    "NIFTY_PHARMA_Return",
    "NIFTY_REALTY_Return"
]

macro_columns = [
    "gold",
    "oil",
    "usd_inr",
    "vix"
]

corr = df[sector_return_columns + macro_columns].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Sector Returns vs Macroeconomic Variables")

plt.show()
```

## **4.2 Sector vs GDELT Correlation**

```python
gdelt_columns = [
    "AvgTone",
    "AvgGoldstein",
    "EventCount",
    "MilitaryCount",
    "ProtestCount",
    "ThreatCount",
    "USA_Conflict",
    "China_Conflict",
    "Russia_Conflict",
    "India_Conflict"
]

corr = df[
    sector_return_columns +
    gdelt_columns
].corr()

plt.figure(figsize=(15,12))

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=True
)

plt.title("Sector Returns vs GDELT Variables")

plt.show()
```

## **9.3 Scatter Plot Analysis**

```python
plt.figure(figsize=(6,5))

sns.scatterplot(
    data=df,
    x="USA_Conflict",
    y="NIFTY50_Return"
)

plt.title("USA Conflict vs NIFTY50 Return")

plt.show()
```

## **4.5 Pairwise Correlation Table**

```python
cross_corr = df[
    sector_return_columns +
    macro_columns +
    gdelt_columns
].corr()

cross_corr.round(2)
```

## **4.6 Cross-Domain Summary**

- India VIX exhibits a consistent negative relationship with sector returns, indicating that increased market uncertainty is generally associated with weaker market performance.
- Gold, Oil, and USD/INR show limited direct linear relationships with daily sector returns, suggesting that their effects may be indirect, nonlinear, or lagged.
- Scatter plots indicate that the relationships between macroeconomic/geopolitical variables and sector returns are not strongly linear, implying that simple correlation analysis may not fully capture their influence.
